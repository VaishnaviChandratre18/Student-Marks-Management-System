STUDENT MARKS MANAGEMENT SYSTEM

```python
students={}
print("--------STUDENT MARK MANAGEMENT SYSTEM---------")
print("1. Add students")
print("2. Display students")
print("3. Show average marks")
print("4. Show top student")
print("5. Goodbye")

while True:
    choice=int(input("Enter the choice:"))
    if choice==1:
        name=input("Enter the name:")
        marks=int(input("Enter the marks:"))
        students[name]=marks
        print("Student Added!")
    elif choice==2:
        print(students)
    elif choice==3:
        if len(students)==0:
            print("No data found")
        else:
            print("Average:",sum(students.values())/len(students))
    elif choice==4:
        if len(students)==0:
            print("No data found")
        else:
            top=max(students,key=students.get)
            print("Top students:",top,students[top])
    elif choice==5:
        print("Goodbye..")
        break
    else:
        print("Enter Valid data")
```
